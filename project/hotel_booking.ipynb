```python
df.groupby('children')['is_canceled'].agg(total='count', canceled='sum', cancel_rate='mean') * [1, 1, 100]
```

```python
cancel_rate = df.groupby('children')['is_canceled'].mean() * 100

plt.figure(figsize=(7,5))
cancel_rate.plot(kind='bar', color='skyblue')

plt.title('Cancellation Rate by Number of Children')
plt.xlabel('Number of Children')
plt.ylabel('Cancellation Rate (%)')
plt.xticks(rotation=0)

for i, v in enumerate(cancel_rate):
    plt.text(i, v + 1, f'{v:.1f}%', ha='center')

plt.show()
```
